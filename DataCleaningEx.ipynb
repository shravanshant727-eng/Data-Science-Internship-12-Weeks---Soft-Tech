```python

df["Peak"]=pd.to_numeric(df["Peak"],errors="coerce")
```

```python
df["Peak"]
```

```python
df["Peak"].fillna(np.mean(df["Peak"]),inplace=True)
```

```python
df["Peak"]
```

```python
df["All Time Peak"] = df["All Time Peak"].astype(str).str.extract(r'(\d+)', expand=False)
df["All Time Peak"] = pd.to_numeric(df["All Time Peak"], errors="coerce")
```

```python
df["All Time Peak"]
```

```python
df.drop_duplicates()
```

```python
df.to_csv("data_cleaned.csv",index=False)
```
